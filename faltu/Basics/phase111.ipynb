```python
import numpy as np
import matplotlib.pyplot as plt 
```

```python
# Data Structure:[restaurant_id,2021,2022,2023,2024]
sales_data= np.array ([
    [1, 15000, 18000, 22000, 25000], # Biryani 
    [2, 12000, 14000, 16000,  19000], #muradabadi 
    [3, 20000, 23000, 26000,  30000],# pizza hub
    [4, 18000, 21000, 24000,  27000],#burger point 
    [5, 16000, 18500, 20500,  23000] #chai point 
])
print("===Zomato Sales Analysis===")
print("\nsales data shape",sales_data.shape)
print("\nSample data for 1st 3 restuarant:\n",sales_data[:3])
print("\nSample data for restuarant without their id :\n",sales_data[:,1:])
```

### total sales per year 


```python
print(np.sum(sales_data, axis=0)) #column wise 

print(np.sum(sales_data, axis=1))#row wise

yearly_sum=np.sum(sales_data[:,1:], axis=1)
print("Yearly sum:",yearly_sum) 
print("Total sum :",np.sum(yearly_sum, axis=0))

min_sale=np.min(sales_data[:,1:],axis=1) # giving from all rows 
print("Minimun Sales axis 1:\n",min_sale)

max_sale=np.max(sales_data[:,1:],axis=0) # giving from all columns 
print("Maximum Sales per year  axis 0:\n",max_sale)

average=np.mean(sales_data[:,1:],axis=1)
print("Average sales per restuarant:\n",average)

```

```python
#cummulative sum 
cum_sum=np.cumsum(sales_data[:,1:],axis=1)
print("Cummalative Sum:\n",cum_sum)

plt.figure(figsize=(8,4))
plt.plot(np.mean(cum_sum,axis=0))
plt.title("Average Cumulative Sales per year ")
plt.xlabel("Year")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```
