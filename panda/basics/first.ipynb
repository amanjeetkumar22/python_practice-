# Missing Data 

```python
ddata={
    'A':[1,2,np.nan,4,5],
    'B':[6,7,8,9,10],
    'C':[11,12,13,np.nan,np.nan],
    'D':[14,np.nan,np.nan,np.nan,15]
}
df=pd.DataFrame(ddata)
df
```

### Finding Missing Data

```python
df.isna()
```

```python
# df.isna().count()
df.isna().sum() ## columnwise selection 
```

### Removing Missing Data 

```python
df.dropna() ## row wise 
```

```python
df.dropna(thresh=3) ## thresh stand for min value 
## minimum utni value honi chahiye 
```

### Filling The Missing Data 

na - column wise 
selection 


row wise -- drop 


```python
df
```

```python
df.fillna(0) ##column wise 
```

```python
## Filling particular data in particular row 
v={'A':0,'B':70,'C':80,'D':100}
df.fillna(value=v)
```

```python
df.fillna(df.mean())
```
